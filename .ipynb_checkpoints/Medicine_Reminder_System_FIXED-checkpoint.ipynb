# Medicine Reminder System – Single Class (MedicineStrip)
### OpenCV + YOLOv11 + Python

**Scope:** Detect ONLY a medicine strip (no multi-medicine classification).
Reminds the user on a hardcoded schedule, confirms intake by requiring the
strip to stay visible for 5 continuous seconds, and logs every successful
intake to a CSV file.

### What this model is trained to detect
- Medicine strip ONLY (`MedicineStrip`)

### What it should NOT detect
Human faces, hands, mobile phones, books, laptops, background objects,
tables, chairs. These should appear only in negative (unlabeled) images,
never as a positive class, so the model learns to ignore them.

## Part 4 — Train YOLOv11 (single class)

````python
device = 0 if torch.cuda.is_available() else "cpu"

model = YOLO("yolo11n.pt")

results = model.train(
    data="data.yaml",
    epochs=50,
    imgsz=512,
    batch=8,
    device=device,
    workers=2,
    cache=False,
    patience=20
)
print("Training finished.")
````

````python
metrics = model.val()
metrics
````

````python
results = model.predict(
    source="test/images",
    save=True,
    conf=0.25
)
print("Predictions saved under runs/detect/predict/")
````

# Part 5 — Real-Time Detection + Reminder System

````python
import cv2
import pyttsx3
import pandas as pd
import os
import time       
from datetime import datetime
from ultralytics import YOLO
````

### 5.1 — Medicine Schedule (hardcoded)

````python
# Add as many entries as needed. Time format: 24-hour "HH:MM"
MEDICINE_SCHEDULE = [
    {
        "name": "Paracetamol",
        "dosage": "1 Tablet",
        "time": "21:00",
        "instruction": "Take after dinner."
    },
]

# Internal state per schedule entry — tracks whether today's dose
# has already been recorded, so we don't repeat once it's done.
for item in MEDICINE_SCHEDULE:
    item["taken_today"] = False
    item["taken_date"] = None       # date string when last marked taken
    item["reminder_active"] = False # True once scheduled time has passed and not yet taken
    item["strip_visible_since"] = None  # timestamp when strip first became continuously visible
    item["last_missed_reminder"] = 0    # timestamp of last "please take your medicine" nag
    item["spoken_initial_reminder"] = False  # the once-only "It is 9 PM..." message
````

### 5.3 — Detection History CSV

````python
CSV_PATH = "medicine_dose_history.csv"
CSV_COLUMNS = [
    "Date", "Time", "TimeZone", "SessionID", "Medicine", "Dose", "Event",
    "ReminderNumber", "PillTakenStatus", "Message", "Confidence", "Box", "EvidenceFile"
]

if not os.path.exists(CSV_PATH):
    pd.DataFrame(columns=CSV_COLUMNS).to_csv(CSV_PATH, index=False)

print("Dose CSV ready ->", CSV_PATH)
````

### 5.4 — Main Loop

State machine per scheduled medicine, each frame:

1. **Before scheduled time:** nothing happens for this entry.
2. **At/after scheduled time, not yet taken today:**
   - First time this triggers -> speak the reminder ONCE
     (`"It is 9 PM. Please take one tablet of Paracetamol after dinner."`)
   - If the strip is detected (conf > 0.70): start/continue a 5-second
     continuous-visibility timer. Losing detection resets the timer.
   - Once the strip has been visible for 5 continuous seconds ->
     speak `"Medicine recorded successfully."`, log to CSV, mark taken
     for today, stop reminding for this entry.
   - If the strip is NOT detected and 5+ minutes have passed since the
     last nag -> speak `"Reminder. Please take your medicine."` and
     repeat every 5 minutes until taken.
3. **Already taken today:** entry is skipped entirely.

Press **q** in the webcam window to quit.


````python
import cv2

cap = cv2.VideoCapture(0)

while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        break

    cv2.imshow("Medicine Reminder", frame)

    key = cv2.waitKey(10) & 0xFF

    if key == ord('q') or key == 27:   # q or Esc
        break

cap.release()
cv2.destroyAllWindows()
````

## Project Summary

- Single-class YOLOv11 detector: `MedicineStrip` only
- Real-time webcam detection via OpenCV with confidence threshold 0.70
- Hardcoded medicine schedule (name, dosage, time, instruction)
- One-time spoken reminder at scheduled time
- 5-second continuous-visibility confirmation before logging intake
- "Medicine recorded successfully" confirmation + CSV logging
  (Date, Time, Medicine, Dose, Status)
- Repeats a missed-dose reminder every 5 minutes until the strip is detected
- Stops reminding once intake is confirmed for the day
