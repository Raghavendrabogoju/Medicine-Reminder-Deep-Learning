### 5. Read Sample Image using OpenCV

```python
import glob
img_path=glob.glob("train/images/*")[0]
img=cv2.imread(img_path)
rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(rgb); plt.axis("off")
print(img_path)
```

### 6. OpenCV Image Processing

```python
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
blur=cv2.GaussianBlur(gray,(5,5),0)
edges=cv2.Canny(blur,50,150)

plt.figure(figsize=(12,4))
plt.subplot(131); plt.imshow(rgb); plt.title("Original"); plt.axis("off")
plt.subplot(132); plt.imshow(gray,cmap="gray"); plt.title("Gray"); plt.axis("off")
plt.subplot(133); plt.imshow(edges,cmap="gray"); plt.title("Edges"); plt.axis("off")
plt.show()
```

### 7. Morphological Operations

```python
kernel=np.ones((3,3),np.uint8)
dilation=cv2.dilate(edges,kernel,iterations=1)
erosion=cv2.erode(edges,kernel,iterations=1)

plt.figure(figsize=(8,4))
plt.subplot(121); plt.imshow(dilation,cmap="gray"); plt.title("Dilation")
plt.subplot(122); plt.imshow(erosion,cmap="gray"); plt.title("Erosion")
plt.show()
```

### 8. Train YOLOv8

```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

results = model.train(
    data="data.yaml",
    epochs=3,
    imgsz=512,
    batch=16,
    device=0,
    workers=8,
    cache=False,
    amp=True
)
```

### 9. Validate Model

```python
metrics=model.val()
metrics
```

### 10. Predict on Test Images

```python
results=model.predict(
    source="test/images",
    save=True,
    conf=0.25
)
```

### 6. Real-Time Webcam Detection

```python
from ultralytics import YOLO

model = YOLO(
    r"C:\Users\HANUMAN NARAYANA\OneDrive\Apps\Desktop\runs\detect\train\weights\best.pt"
)

print("✅ Model loaded successfully")
```

```python
import cv2
import pyttsx3
from ultralytics import YOLO

model = YOLO( r"C:\Users\HANUMAN NARAYANA\OneDrive\Apps\Desktop\runs\detect\train\weights\best.pt")

engine = pyttsx3.init()

spoken = False

cap = cv2.VideoCapture(0, cv2.CAP_DSHOW)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    results = model(frame, verbose=False)

    annotated = results[0].plot()

    detected = False

    for box in results[0].boxes:
        conf = float(box.conf)

        if conf > 0.70:
            detected = True

    if detected and not spoken:
        engine.say("Please take your medicine.")
        engine.runAndWait()
        spoken = True

    if not detected:
        spoken = False

    cv2.imshow("Medicine Reminder", annotated)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

### 8. Simple Reminder Logic

```python
medicine_time="21:00"

current=datetime.now().strftime("%H:%M")

if current==medicine_time:
    engine.say("It is time to take your medicine.")
    engine.runAndWait()
    print("Reminder Triggered")
else:
    print("Current Time:",current)
    print("Reminder Time:",medicine_time)
```
