### 2. Import Libraries

```
import cv2
import pyttsx3
import pandas as pd
from ultralytics import YOLO
from datetime import datetime
import os,time
```

### 3. Initialize Voice Engine

```
engine=pyttsx3.init()
engine.setProperty("rate",160)
print("Voice engine ready")
```

### 4. Load Trained Model

```
model=YOLO("MedicineProject/MedicineDetector/weights/best.pt")
print("Model Loaded")
```

### 5. Create Detection Log

```
if not os.path.exists("detection_history.csv"):
    pd.DataFrame(columns=["Date","Time","Object","Confidence"]).to_csv("detection_history.csv",index=False)
print("CSV Ready")
```

### 6. Real-Time Webcam Detection

```
cap=cv2.VideoCapture(0)

spoken=False

while True:

    ret,frame=cap.read()
    if not ret:
        break

    results=model(frame,verbose=False)

    for r in results:
        boxes=r.boxes

        for box in boxes:

            x1,y1,x2,y2=map(int,box.xyxy[0])

            conf=float(box.conf[0])

            cls=int(box.cls[0])

            name=model.names[cls]

            cv2.rectangle(frame,(x1,y1),(x2,y2),(0,255,0),2)

            cv2.putText(frame,
                        f"{name} {conf:.2f}",
                        (x1,y1-10),
                        cv2.FONT_HERSHEY_SIMPLEX,
                        0.7,
                        (0,255,0),
                        2)

            if conf>0.70:

                if not spoken:
                    engine.say("Medicine Detected")
                    engine.runAndWait()
                    spoken=True

                now=datetime.now()

                df=pd.read_csv("detection_history.csv")

                df.loc[len(df)]=[
                    now.strftime("%Y-%m-%d"),
                    now.strftime("%H:%M:%S"),
                    name,
                    round(conf,2)
                ]

                df.to_csv("detection_history.csv",index=False)

    current=datetime.now().strftime("%d-%m-%Y %H:%M:%S")

    cv2.putText(frame,
                current,
                (10,30),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (255,255,0),
                2)

    cv2.imshow("Medicine Reminder System",frame)

    key=cv2.waitKey(1)

    if key==ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

### 7. View Detection History

```
pd.read_csv("detection_history.csv").tail()
```

### 8. Simple Reminder Logic

```
medicine_time="21:00"

current=datetime.now().strftime("%H:%M")

if current==medicine_time:
    engine.say("It is time to take your medicine.")
    engine.runAndWait()
    print("Reminder Triggered")
else:
    print("Current Time:",current)
    print("Reminder Time:",medicine_time)
```
