# Medicine Reminder System – Single Class (MedicineStrip)
### OpenCV + YOLOv11 + Python

**Scope:** Detect ONLY a medicine strip (no multi-medicine classification).
Reminds the user on a hardcoded schedule, confirms intake by requiring the
strip to stay visible for 5 continuous seconds, and logs every successful
intake to a CSV file.


### What this model is trained to detect
- Medicine strip ONLY (`MedicineStrip`)

### What it should NOT detect
Human faces, hands, mobile phones, books, laptops, background objects,
tables, chairs. These should appear only in negative (unlabeled) images,
never as a positive class, so the model learns to ignore them.

## Part 4 — Train YOLOv11 (single class)

````python
model = YOLO("yolo11n.pt")  # nano model — fast, good enough for 1 class

results = model.train(
    data="data.yaml",
    epochs=50,          # small dataset -> more epochs helps, watch for overfitting
    imgsz=512,
    batch=8,
    device=0,            # set to 'cpu' if no GPU available
    workers=4,
    cache=False,
    amp=True,
    patience=20
)
````

````python
metrics = model.val()
metrics
````

# Part 5 — Real-Time Detection + Reminder System

````python
import cv2
import pyttsx3
import pandas as pd
import os
import time
from datetime import datetime
from ultralytics import YOLO
````

### 5.1 — Medicine Schedule (hardcoded)

````python
# Add as many entries as needed. Time format: 24-hour "HH:MM"
MEDICINE_SCHEDULE = [
    {
        "name": "Paracetamol",
        "dosage": "1 Tablet",
        "time": "21:00",
        "instruction": "Take after dinner."
    },
]

# Internal state per schedule entry — tracks whether today's dose
# has already been recorded, so we don't repeat once it's done.
for item in MEDICINE_SCHEDULE:
    item["taken_today"] = False
    item["taken_date"] = None       # date string when last marked taken
    item["reminder_active"] = False # True once scheduled time has passed and not yet taken
    item["strip_visible_since"] = None  # timestamp when strip first became continuously visible
    item["last_missed_reminder"] = 0    # timestamp of last "please take your medicine" nag
    item["spoken_initial_reminder"] = False  # the once-only "It is 9 PM..." message
````

### 5.2 — Load Model and Voice Engine

````python
MODEL_PATH = "best.pt"   # update path if your weights live elsewhere
CONF_THRESHOLD = 0.70
CONFIRM_SECONDS = 5          # strip must stay visible this long to confirm intake
MISSED_REMINDER_INTERVAL = 300  # 5 minutes, in seconds

model = YOLO(MODEL_PATH)
engine = pyttsx3.init()
engine.setProperty("rate", 160)

print("Model and voice engine ready.")
````

### 5.3 — Detection History CSV

````python
CSV_PATH = "detection_history.csv"

if not os.path.exists(CSV_PATH):
    pd.DataFrame(columns=["Date", "Time", "Medicine", "Dose", "Status"]).to_csv(CSV_PATH, index=False)

def log_intake(medicine_name, dosage, status="Taken"):
    now = datetime.now()
    row = pd.DataFrame(
        [[now.strftime("%Y-%m-%d"), now.strftime("%H:%M:%S"), medicine_name, dosage, status]],
        columns=["Date", "Time", "Medicine", "Dose", "Status"]
    )
    row.to_csv(CSV_PATH, mode="a", header=False, index=False)

print("CSV logging ready ->", CSV_PATH)
````

### 5.4 — Main Loop

State machine per scheduled medicine, each frame:

1. **Before scheduled time:** nothing happens for this entry.
2. **At/after scheduled time, not yet taken today:**
   - First time this triggers -> speak the reminder ONCE
     (`"It is 9 PM. Please take one tablet of Paracetamol after dinner."`)
   - If the strip is detected (conf > 0.70): start/continue a 5-second
     continuous-visibility timer. Losing detection resets the timer.
   - Once the strip has been visible for 5 continuous seconds ->
     speak `"Medicine recorded successfully."`, log to CSV, mark taken
     for today, stop reminding for this entry.
   - If the strip is NOT detected and 5+ minutes have passed since the
     last nag -> speak `"Reminder. Please take your medicine."` and
     repeat every 5 minutes until taken.
3. **Already taken today:** entry is skipped entirely.

Press **q** in the webcam window to quit.


````python
def get_today_str():
    return datetime.now().strftime("%Y-%m-%d")

def reset_if_new_day(item):
    today = get_today_str()
    if item["taken_date"] != today:
        item["taken_today"] = False
        item["taken_date"] = None
        item["reminder_active"] = False
        item["strip_visible_since"] = None
        item["last_missed_reminder"] = 0
        item["spoken_initial_reminder"] = False

def run_reminder_system():
    cap = cv2.VideoCapture(0, cv2.CAP_DSHOW)

    if not cap.isOpened():
        print("Could not open webcam.")
        return

    print("Webcam started. Press 'q' to quit.")

    while True:
        ret, frame = cap.read()
        if not ret:
            print("Webcam read failed.")
            break

        results = model(frame, verbose=False)
        annotated = frame.copy()

        # Find the best medicine-strip detection in this frame
        best_conf = 0.0
        strip_detected = False
        for box in results[0].boxes:
            conf = float(box.conf[0])
            if conf > best_conf:
                best_conf = conf
            if conf > CONF_THRESHOLD:
                strip_detected = True
                x1, y1, x2, y2 = map(int, box.xyxy[0])
                cv2.rectangle(annotated, (x1, y1), (x2, y2), (0, 255, 0), 2)
                label = f"Medicine Strip {conf*100:.0f}%"
                cv2.putText(annotated, label, (x1, max(20, y1 - 10)),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

        now = datetime.now()
        current_time_str = now.strftime("%H:%M")
        now_ts = time.time()

        for item in MEDICINE_SCHEDULE:
            reset_if_new_day(item)

            if item["taken_today"]:
                continue  # already done for today, nothing to do

            # Has the scheduled time arrived (or passed) today?
            if current_time_str >= item["time"]:
                item["reminder_active"] = True

            if not item["reminder_active"]:
                continue  # not time yet for this medicine

            # Speak the initial reminder exactly once
            if not item["spoken_initial_reminder"]:
                hour_12 = datetime.strptime(item["time"], "%H:%M").strftime("%I %p").lstrip("0")
                message = (f"It is {hour_12}. Please take {item['dosage']} of "
                           f"{item['name']} {item['instruction'].lower()}")
                print("SPEAKING:", message)
                engine.say(message)
                engine.runAndWait()
                item["spoken_initial_reminder"] = True
                item["last_missed_reminder"] = now_ts  # avoid an immediate duplicate nag

            if strip_detected:
                if item["strip_visible_since"] is None:
                    item["strip_visible_since"] = now_ts

                visible_duration = now_ts - item["strip_visible_since"]
                cv2.putText(annotated, f"Confirming: {visible_duration:.1f}s / {CONFIRM_SECONDS}s",
                            (10, 70), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 200, 255), 2)

                if visible_duration >= CONFIRM_SECONDS:
                    print("SPEAKING: Medicine recorded successfully.")
                    engine.say("Medicine recorded successfully.")
                    engine.runAndWait()
                    log_intake(item["name"], item["dosage"], status="Taken")
                    item["taken_today"] = True
                    item["taken_date"] = get_today_str()
                    item["reminder_active"] = False
                    item["strip_visible_since"] = None
            else:
                # Strip not visible -> reset the confirmation timer
                item["strip_visible_since"] = None

                # Nag every 5 minutes until taken
                if now_ts - item["last_missed_reminder"] >= MISSED_REMINDER_INTERVAL:
                    print("SPEAKING: Reminder. Please take your medicine.")
                    engine.say("Reminder. Please take your medicine.")
                    engine.runAndWait()
                    item["last_missed_reminder"] = now_ts

        status_text = "Strip detected" if strip_detected else "No strip in frame"
        cv2.putText(annotated, status_text, (10, 30),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0) if strip_detected else (0, 0, 255), 2)
        cv2.putText(annotated, datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
                    (10, annotated.shape[0] - 15), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 1)

        cv2.imshow("Medicine Reminder System", annotated)

        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()
    print("Webcam closed.")
````

````python
# Run the system
run_reminder_system()
````

## Project Summary

- Single-class YOLOv11 detector: `MedicineStrip` only
- Real-time webcam detection via OpenCV with confidence threshold 0.70
- Hardcoded medicine schedule (name, dosage, time, instruction)
- One-time spoken reminder at scheduled time
- 5-second continuous-visibility confirmation before logging intake
- "Medicine recorded successfully" confirmation + CSV logging
  (Date, Time, Medicine, Dose, Status)
- Repeats a missed-dose reminder every 5 minutes until the strip is detected
- Stops reminding once intake is confirmed for the day
